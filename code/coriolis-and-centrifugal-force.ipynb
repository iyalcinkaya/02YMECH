```python
import matplotlib.pyplot as plt
import numpy as np

o = 5      # Angular velocity of carousel
R = 1      # Radius of carousel

r = [0, 0]     # Position of the ball
v = [0, 1] # Velocity of ball
a = [0, 0]     # Acceleration of ball

dt = 0.01  # Time step
t = 0      # Time

def acceleration(r, v, t):
    return [2*o*v[1] + o*o*r[0], -2*o*v[0] + o*o*r[1]]

def velocity(v, a, dt):
    return [v[0] + a[0]*dt, v[1] + a[1]*dt]

def position(r, v, dt):
    return [r[0] + v[0]*dt, r[1] + v[1]*dt]

# Can you make the plot ratio 1:1?
plt.axes().set_aspect('equal')

while t < 100:
    a = acceleration(r, v, dt)
    v = velocity(v, a, dt)
    r = position(r, v, dt)
    t += dt
    plt.plot(r[0], r[1],'bo', markersize=1)
    if r[0]*r[0] + r[1]*r[1] > R*R:
        break

plt.xlim(-R, R)
plt.ylim(-R, R)

# Involve a circle to show the boundary of the carousel
x = np.linspace(-R, R, 100)
y = np.sqrt(R*R - x*x)
plt.plot(x, y, 'r')
plt.plot(x, -y, 'r')
```
